```python
df.drop(
    columns=[
        "order_id",
        "customer_id",
        "product_id",
        "order_date",
        "order_time",
        "customer_name_x",
        "product_name",
        "customer_city_x",
        "customer_state_x",
        "customer_country_x",
        "customer_postal_code_x",
        "currency",
        "return_status",
        "return_reason",
        "customer_review",
        "campaign_name",
        "coupon_code",
        "is_repeat_customer",
        "customer_name_y",
        "customer_city_y",
        "customer_state_y",
        "customer_country_y",
        "customer_postal_code_y",
        "customer_acquisition_cost",
        "payment_status"
    ],
    inplace=True,
)
```

```python
import pandas as pd

from sklearn.compose import ColumnTransformer 
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer

oneHot_column = [
    "sales_channel",
    "region_x",
    "supplier",
    "region_y"
]

label_column = [
    "gender_x",
    "customer_segment_x",
    "customer_type",
    "brand",
    "product_subcategory",
    "product_category",
    "payment_method",
    "shipping_method",
    "warehouse",
    "delivery_status",
    "review_sentiment",
    "marketing_channel",
    "gender_y",
    "customer_segment_y",
]

ct = ColumnTransformer(
    [
        (
            "one-hot-encoder",
            OneHotEncoder(
                sparse_output=False,
                dtype=int,
                handle_unknown="ignore"
            ),
            oneHot_column,
        ),
        (
            "ordinal-encoder",
            OrdinalEncoder(
                handle_unknown="use_encoded_value",
                unknown_value=-1
            ),
            label_column,
        ),
    ],
    remainder="passthrough",
    verbose_feature_names_out=False
)

df_transformed = ct.fit_transform(df)

feature_names = ct.get_feature_names_out()

df = pd.DataFrame(
    df_transformed,
    columns=feature_names,
    index=df.index
)

impute = SimpleImputer(strategy="mean")

df_transformed = impute.fit_transform(df)

df = pd.DataFrame(
    df_transformed,
    columns=feature_names,
    index=df.index
)
```

```python
x = df.drop(columns=["order_status"])
y = df["order_status"]
```

```python
x.columns
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

x = scaler.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
from xgboost import XGBClassifier

model = XGBClassifier(
    n_estimators=2500,
    learning_rate=0.05,
    max_depth=7,
    random_state=42,
)

model.fit(x_train , y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import classification_report , confusion_matrix

report = classification_report(y_test , y_pred)
confusion = confusion_matrix(y_test , y_pred)

print("report ===> \n" , report)
print("=======================")
print("confusion ===> \n" , confusion)
```
