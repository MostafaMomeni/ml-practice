# -import

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# -load Data

```python
df = pd.read_csv("data.csv")
df.head()
```

# -clean Data

```python
df.info()
```

## one hot encoder

```python
from sklearn.preprocessing import OneHotEncoder

oneHotEncoder_column = []

oh = OneHotEncoder(sparse_output=False)

new_df = oh.fit_transform(df[oneHotEncoder_column])
new_df = pd.DataFrame(new_df , columns = oh.get_feature_names_out(oneHotEncoder_column))
df = pd.concat([df , new_df] , axis = 1)
df.drop(oneHotEncoder_column , axis = 1 , inplace = True)
```

## label encoder

```python
from sklearn.preprocessing import LabelEncoder

labeled_columns = []

for col in labeled_columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
```

## date cleaning

```python
# -----------------------------------
# 1. تبدیل ستون‌ها به تاریخ
# -----------------------------------

df["start_date"] = pd.to_datetime(
    df["start_date"],
    errors="coerce"
)

df["end_date"] = pd.to_datetime(
    df["end_date"],
    errors="coerce"
)


# -----------------------------------
# 2. مدت زمان بین تاریخ شروع و پایان
# -----------------------------------

# تعداد روز
df["duration_days"] = (
    df["end_date"] - df["start_date"]
).dt.days


# تعداد ماه تقریبی
df["duration_months"] = (
    df["duration_days"] / 30.44
)

# اگر ماه را به صورت عدد صحیح می‌خواهی:
df["duration_months"] = df["duration_months"].round().astype("Int64")


# تعداد سال تقریبی
df["duration_years"] = (
    df["duration_days"] / 365.25
)

df["duration_years"] = df["duration_years"].round().astype("Int64")


# -----------------------------------
# 3. ویژگی‌های مربوط به تاریخ شروع
# -----------------------------------

df["start_year"] = df["start_date"].dt.year
df["start_month"] = df["start_date"].dt.month
df["start_week"] = df["start_date"].dt.isocalendar().week
df["start_day"] = df["start_date"].dt.day

# فصل
df["start_quarter"] = df["start_date"].dt.quarter


# -----------------------------------
# 4. ویژگی‌های مربوط به تاریخ پایان
# -----------------------------------

df["end_year"] = df["end_date"].dt.year
df["end_month"] = df["end_date"].dt.month
df["end_week"] = df["end_date"].dt.isocalendar().week
df["end_day"] = df["end_date"].dt.day

# فصل
df["end_quarter"] = df["end_date"].dt.quarter
```

## x and y

```python
x = df.drop(["target"] , axis=1)
y = df["target"]
```

## pca

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=20 , random_state=42)

x = pca.fit_transform(x)
```

## scale

```python
from sklearn.preprocessing import MinMaxScaler , StandardScaler , RobustScaler

scaler = MinMaxScaler()

x = scaler.fit_transform(x)
```

## split data

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , random_state=42 , test_size=0.3)
```

# -model

## regression

```python
from xgboost import XGBRegressor

model = XGBRegressor(
    n_estimators = 1500,
    learning_rate = 0.05,
    max_depth = 7,
)

model.fit(x_train , y_train)
```

```python
from sklearn.ensemble import (
    RandomForestRegressor,
)

model_1 = RandomForestRegressor(
    n_estimators=500, max_depth=7, random_state=42, n_jobs=-1
)

model.fit(x_train , y_train)
```

### ensemble model

```python
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor,
    StackingRegressor,
)
from xgboost import XGBRegressor
from sklearn.linear_model import Ridge

model_1 = RandomForestRegressor(
    n_estimators=500, max_depth=7, random_state=42, n_jobs=-1
)

model_2 = XGBRegressor(
    n_estimators=1500,
    learning_rate=0.05,
    max_depth=7,
    random_state=42,
)
model_3 = GradientBoostingRegressor(
    n_estimators=1500,
    learning_rate=0.05,
    max_depth=7,
    random_state=42,
)


# =========================
# 5. ترکیب مدل‌ها
# =========================

model = StackingRegressor(
    estimators=[
        ("random_forest", model_1),
        ("xgboost", model_2),
        ("GradientBoosting", model_3),
    ],
    final_estimator=Ridge(alpha=1.0),
    cv=5,
    n_jobs=-1,
)

model.fit(x_train , y_train)
```

## classification

```python
from xgboost import XGBClassifier

model = XGBClassifier(
    n_estimators = 4000,
    learning_rate = 0.05,
    max_depth = 7,
)

model.fit(x_train , y_train)
```

```python
from sklearn.ensemble import GradientBoostingClassifier

model = GradientBoostingClassifier(
    n_estimators = 3500,
    max_depth = 5,
    random_state=42
)

model.fit(x_train , y_train)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators = 3500,
    max_depth = 5,
    random_state=42
)

model.fit(x_train , y_train)
```

### ensemble model

```python
from sklearn.ensemble import (
    RandomForestClassifier,
    StackingClassifier
)
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from sklearn.svm import SVC

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    random_state=42,
    eval_metric="logloss"
)

svm = SVC(
    probability=True,
    random_state=42
)


meta_model = LogisticRegression(
    max_iter=1000
)


# Stacking
stacking_model = StackingClassifier(
    estimators=[
        ("rf", rf),
        ("xgb", xgb),
        ("svm", svm)
    ],
    final_estimator=meta_model,
    cv=5,
    n_jobs=-1
)


stacking_model.fit(
    x_train,
    y_train
)

```

## clustering

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)


# آموزش
kmeans.fit(x_train)


# شماره گروه هر رکورد
df["kmeans_cluster"] = kmeans.labels_


# -----------------------------
# ارزیابی
# -----------------------------

score = silhouette_score(
    x_train,
    kmeans.labels_
)

print("Silhouette Score:", score)
```

```python
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score

dbscan = DBSCAN(
    eps=0.5,
    min_samples=5
)


dbscan.fit(x_train)


# اضافه کردن گروه
df["dbscan_cluster"] = dbscan.labels_


# -----------------------------
# تعداد داده‌های هر گروه
# -----------------------------

print(
    df["dbscan_cluster"].value_counts()
)
```

# -metrics

## regression

```python
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score , root_mean_squared_error

mae = mean_absolute_error(y_test , y_pred)
mse = mean_squared_error(y_test , y_pred)
rmse = root_mean_squared_error(y_test , y_pred)
r2 = r2_score(y_test , y_pred)

print("mae" , mae)
print("mse" , mse)
print("rmse" , rmse)
print("r2" , r2)
```

## classification

```python
from sklearn.metrics import classification_report , confusion_matrix

report = classification_report(y_test , y_pred)
confusion = confusion_matrix(y_test , y_pred)

print(report)
print("========")
print(confusion)
```
